```python
import sklearn  
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.metrics import precision_score, recall_score, f1_score
from sklearn.metrics import RocCurveDisplay
```

```python
np.random.seed(40)
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2)#20% to test and 80% to train the model
```

```python
X_train.head()
```

```python
# Check for missing values in your training set
X_train.isna().sum()
```

```python
# Fill missing values with the median of each column
X_train = X_train.fillna(X_train.median())
X_test = X_test.fillna(X_test.median()) # Do the same for test set
```

```python
X_train.head()
Y_train, len(Y_train)
```

```python
models={"KNN":KNeighborsClassifier(),
        "Logistic Regression":LogisticRegression(),
        "Random Forest":RandomForestClassifier()}
def fit_and_score(models, X_train, X_test, Y_train, Y_test):
    np.random.seed(42)
    model_scores={}
    for name,model in models.items():
        model.fit(X_train,Y_train)
        model_scores[name]=model.score(X_test,Y_test)   
    return model_scores
    # model_scores=fit_and_score(models=models,
    #                        X_train=X_train,
    #                        X_test=X_test,
    #                        Y_train=Y_train,
    #                        Y_test=Y_test) 
    # model_scores   
```

```python
model_scores = fit_and_score(models=models,
                           X_train=X_train,
                           X_test=X_test,
                           Y_train=Y_train,
                           Y_test=Y_test) 
model_scores   
```

```python
model_compare = pd.DataFrame(model_scores, index=["accuracy"])
model_compare.T.plot.bar();
#logistic regression is the best model for this dataset with accuracy of 85%
```

```python

train_scores = []


test_scores = []

neighbors = range(1, 21) # 1 to 20

knn = KNeighborsClassifier()

# Loop through different neighbors values
for i in neighbors:
    knn.set_params(n_neighbors = i)
    
    # Fit the algorithm
    knn.fit(X_train, Y_train)
    
    # Update the training scores
    train_scores.append(knn.score(X_train, Y_train))
    
    # Update the test scores
    test_scores.append(knn.score(X_test, Y_test))
```

```python
train_scores
```

VISUALIZATION

```python
plt.plot(neighbors, train_scores, label="Train score")
plt.plot(neighbors, test_scores, label="Test score")
plt.xticks(np.arange(1, 21, 1))
plt.xlabel("Number of neighbors")
plt.ylabel("Model score")
plt.legend()

print(f"Maximum KNN score on the test data: {max(test_scores)*100:.2f}%")
```

```python
# Different LogisticRegression hyperparameters
log_reg_grid = {"C": np.logspace(-4, 4, 20),
                "solver": ["liblinear"]}

#  RandomForestClassifier hyperparameters
rf_grid = {"n_estimators": np.arange(10, 1000, 50),
           "max_depth": [None, 3, 5, 10],
           "min_samples_split": np.arange(2, 20, 2),
           "min_samples_leaf": np.arange(1, 20, 2)}
```

```python
%%time

# Setup random seed
np.random.seed(42)

# Setup random hyperparameter search for LogisticRegression
rs_log_reg = RandomizedSearchCV(LogisticRegression(),
                                param_distributions=log_reg_grid,
                                cv=5,
                                n_iter=20,
                                verbose=True)

# Fit random hyperparameter search model
rs_log_reg.fit(X_train, Y_train);
```

```python
rs_log_reg.score(X_test, Y_test)
```

```python
%%time 

# Setup random seed
np.random.seed(42)

# Setup random hyperparameter search for RandomForestClassifier
rs_rf = RandomizedSearchCV(RandomForestClassifier(),
                           param_distributions=rf_grid,
                           cv=5,
                           n_iter=20,
                           verbose=True)

# Fit random hyperparameter search model
rs_rf.fit(X_train, Y_train);
```

```python
# Find the best parameters
rs_rf.best_params_
```

```python
# Evaluating the randomized search random forest model
rs_rf.score(X_test, Y_test)
```
